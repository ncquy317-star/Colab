```python
import matplotlib.pyplot as plt
import numpy as np

def cot(x):
    return 1 / np.tan(x)

# Generate x values, avoiding the exact points of discontinuity initially for plotting continuous segments
x = np.linspace(0.01, 2 * np.pi - 0.01, 500) # Avoid 0 and 2pi for initial linspace

# Create a more refined x array to clearly show asymptotes
x_plot = np.array([])
for i in range(3):
    segment_start = i * np.pi + 0.01
    segment_end = (i + 1) * np.pi - 0.01
    if segment_start < 2 * np.pi and segment_end <= 2 * np.pi:
        x_segment = np.linspace(segment_start, segment_end, 200)
        x_plot = np.concatenate((x_plot, x_segment, [np.nan])) # Add NaN to break the line at asymptotes

y_plot = cot(x_plot)

plt.figure(figsize=(10, 6))
plt.plot(x_plot, y_plot, label='f(x) = cot(x)', color='blue')

# Plot vertical asymptotes as dashed red lines
asymptotes = [0, np.pi, 2 * np.pi]
for val in asymptotes:
    plt.axvline(x=val, color='red', linestyle='--', label=f'Vertical Asymptote at x={val/np.pi}\u03c0' if val!=0 and val!=2*np.pi else f'Vertical Asymptote at x={val}')

plt.ylim(-10, 10) # Limit y-axis for better visualization of asymptotes
plt.xlim(0, 2 * np.pi) # Set x-axis limit to the interval [0, 2pi]
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Graph of f(x) = cot(x) on [0, 2π] with Discontinuities')
plt.xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi],
           ['0', r'$π/2$', r'$π$', r'$3π/2$', r'$2π$'])
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define the function
def f(x):
  return 4 * np.exp(x - 3) - 2 * (x**3 - 5*x + 9)

# Generate x values
x_values = np.linspace(-5, 5, 500) # Adjusted range to better show the function's behavior

# Calculate y values
y_values = f(x_values)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='f(x) = 4e^(x-3) - 2(x^3 - 5x + 9)', color='purple')

# Add labels and title
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Graph of f(x) = 4e^(x-3) - 2(x^3 - 5x + 9)')

# Add grid and legend
plt.grid(True)
plt.legend()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define the function
def f(x):
  return x**2 - 2*x - 9 - 3*np.sin(x)

# Generate x values
x_values = np.linspace(-5, 10, 500) # Adjust the range as needed

# Calculate y values
y_values = f(x_values)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='f(x) = x^2 - 2x - 9 - 3sin(x)', color='blue')

# Add labels and title
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Graph of f(x) = x^2 - 2x - 9 - 3sin(x)')

# Add grid and legend
plt.grid(True)
plt.legend()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Tạo dữ liệu cho 3 khoảng
x1 = np.linspace(-5, 0, 100)
y1 = 3 * x1 - 10

x2 = np.linspace(0, 2, 100)
y2 = x2**2 + 3 * x2 - 10

x3 = np.linspace(2, 6, 100)
y3 = 3 * x3 - 6

# Vẽ đồ thị
plt.figure(figsize=(10, 6))

# Vẽ 3 nhánh hàm số
plt.plot(x1, y1, label=r'$3x - 10 \quad (x \leq 0)$', color='blue', linewidth=2)
plt.plot(x2, y2, label=r'$x^2 + 3x - 10 \quad (0 < x < 2)$', color='red', linewidth=2)
plt.plot(x3, y3, label=r'$3x - 6 \quad (x \geq 2)$', color='green', linewidth=2)

# Đánh dấu điểm nghiệm c = 1, f(c) = -6
plt.scatter([1], [-6], color='black', s=150, zorder=5, label='Nghiệm c=1')
# Vẽ đường gióng nét đứt tại y = -6
plt.axhline(y=-6, color='gray', linestyle='--', alpha=0.7)
plt.text(6.2, -6, 'y = -6', verticalalignment='center')

# Trang trí đồ thị
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.title('Đồ thị hàm số từng phần f(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.show()
```
